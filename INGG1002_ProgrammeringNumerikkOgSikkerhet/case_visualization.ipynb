## Problem Statement

We want to visualize the quadratic equation in Python:

$f(x) = ax^2 + bx + c$

where (a), (b), and (c) are constants. The solutions for (x) are given by the quadratic formula:

$x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}$

$discriminant = b^2 - 4ac$

- discriminant = 0: We have only one solution
- discriminant > 0: We have two solutions
- discriminant < 0: We have no solutions (complex numbers)

$x_{1} = \frac{-b + \sqrt{discriminant}}{2a}$

$x_{2} = \frac{-b - \sqrt{discriminant}}{2a}$

User defined functions:

- quadratic_solver(a, b, c)


```python
# define the quadratic function
def quadratic(x, a, b, c):
    """
    Quadratic function: f(x) = ax^2 + bx + c
    Args:
        x (float): The input value for the quadratic function.
    returns:
        float: The value of the quadratic function at x.
    """
    return a * x**2 + b * x + c


# To visualize the roots of the quadratic function, we can use the function from previous lecture:


def quadratic_solver(a, b, c):
    """
    Solves the quadratic equation ax^2 + bx + c = 0.
    Args:
        a (float): Coefficient of x^2.
        b (float): Coefficient of x.
        c (float): Constant term.

    Returns:
        tuple: A tuple containing the roots of the quadratic equation.
    """

    # Calculate the discriminant
    discriminant = b**2 - 4 * a * c

    if discriminant > 0:
        # Calculate the two distinct real roots
        root_1 = (-b + sqrt(discriminant)) / (2 * a)
        root_2 = (-b - sqrt(discriminant)) / (2 * a)
        return root_1, root_2
    elif discriminant == 0:
        # Calculate the single real root
        root = -b / (2 * a)
        return (root,)
    else:
        # Calculate the complex roots
        real_part = -b / (2 * a)
        imaginary_part = sqrt(-discriminant) / (2 * a)
        root_1 = complex(real_part, imaginary_part)
        root_2 = complex(real_part, -imaginary_part)
        return root_1, root_2
```

## Solution 2: Use of while-loop


```python
x_start = -10
x_end = 10
step = 1

x_values = []
y_values = []

x = x_start
while x <= x_end:
    y = quadratic(x, a, b, c)
    x_values.append(x)
    y_values.append(y)
    print(f"The value of the quadratic function at x={x} is: {y}")
    x = x + step

# Plotting the quadratic function
plt.plot(x_values, y_values, label="Quadratic Function", color="blue")

# Plotting the roots
plt.plot(roots[0], 0, "ro", label="Solution 1")
plt.plot(roots[1], 0, "go", label="Solution 2")

plt.title("Quadratic Function Plot")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()
plt.legend()
plt.show()
```

## Solution 4: Iterative Approach (Bisection method)


```python
def bisection(a_coef, b_coef, c_coef, left, right, tol=1e-6):

    iteration = 0

    while abs(right - left) > tol:

        iteration += 1

        # Find the middle
        mid = (left + right) / 2

        # Calculate function values
        f_left = quadratic(left, a_coef, b_coef, c_coef)
        f_mid = quadratic(mid, a_coef, b_coef, c_coef)

        print(
            f"Iteration {iteration}: "
            f"left = {left:.6f}, "
            f"right = {right:.6f}, "
            f"mid = {mid:.6f}, "
            f"f(mid) = {f_mid:.6f}"
        )

        # If the midpoint is exactly a root
        if f_mid == 0:
            return mid

        # Decide which half contains the root
        if f_left * f_mid < 0:
            right = mid
        else:
            left = mid

    return (left + right) / 2
```

```python
bisection_roots = []

for i in range(len(x_values) - 1):

    y1 = y_values[i]
    y2 = y_values[i + 1]

    # Case 1: x_values[i] is exactly a root
    if y1 == 0.0:
        bisection_roots.append(x_values[i])

    # Case 2: root lies between the two points
    elif y1 * y2 < 0.0:

        print(
            f"\nStarting bisection:",
            f"\n    y1 = {y1:.6f} at x = {x_values[i]:.6f}",
            f"\n    y2 = {y2:.6f} at x = {x_values[i + 1]:.6f}",
        )
        print("------------------")

        root = bisection(a, b, c, x_values[i], x_values[i + 1])

        bisection_roots.append(root)
```

```python
print("Bisection roots:")

for root in bisection_roots:
    print(f"x = {root:.6f}")

print(
    f"Bisection roots of {a}x^2 + {b}x + {c} = 0 are:",
    f"\n    {bisection_roots[0]:.6f}, {bisection_roots[1]:.6f}",
)
```
